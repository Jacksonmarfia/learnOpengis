```python
# 导入需要的模块并设置工作区

import os
from osgeo import ogr

os.chdir("D:/learnpy/learnOpengis-1/Exercise3-todeliver")

inshp = "./data-Exercise3/sites.shp"
datasource = ogr.Open(inshp)
```

```python
# 获得shp驱动
# 获得shp图层
driver = datasource.GetDriver()
layer_shp = datasource.GetLayer(0)
```

```python
# 注册img并打开img
from osgeo import ogr, gdal
from osgeo.gdalconst import *
import numpy
import sys
# register all of the GDAL drivers
gdal.AllRegister()

# open the image
ds = gdal.Open('./data-Exercise3/aster.img', GA_ReadOnly)
if ds is None:
    print('Could not open aster.img')
    sys.exit(1)

```

```python
# 获取所有要素的ID
featureIDlist = []
layer_shp.ResetReading()
feature_shp = layer_shp.GetNextFeature()
while feature_shp:
    featureID = feature_shp.GetField('id')
    featureIDlist.append(featureID)
    feature_shp = layer_shp.GetNextFeature()
```

```python
geo = ds.GetGeoTransform()
```

```python
datasource.GetGeoTransform()
```

```python
originX=geo[0] 
originY=geo[3] 
pixelWidth=geo[1] 
pixelHeight=geo[5]
```

```python
# 获取所有要素的空间坐标并计算像素偏移
X = []
Y = []
layer_shp.ResetReading()
feature_shp = layer_shp.GetNextFeature()
while feature_shp:
    geom = feature_shp.GetGeometryRef()
    f_x = geom.GetX()
    f_y = geom.GetY()

    # 计算像素偏移
    pixel_x = int((f_x - originX) / (pixelWidth))
    pixel_y = int((f_y - originY) / (pixelHeight))

    X.append(pixel_x)
    Y.append(pixel_y)
    feature_shp = layer_shp.GetNextFeature()
```

```python
X
```

```python
# 循环每个波段并获取波段在像素偏移点的值
values = []
for i in range(ds.RasterCount):
    band = ds.GetRasterBand(i + 1)
    for j in range(len(X)):
        value = band.ReadAsArray(X[j], Y[j], 1, 1)
        values.append(value)
```
