```python
driver = gdal.GetDriverByName( 'GTiff' )
dst_filename = 'x_tmp.tif'
dst_ds = driver.Create(dst_filename, xsize=512, ysize=512, bands=1, eType=gdal.GDT_Byte)
```

```python
import numpy
from osgeo import osr
dst_ds.SetGeoTransform( [ 444720, 30, 0, 3751320, 0, -30 ] )
srs = osr.SpatialReference()
srs.SetUTM(11,True )
srs.SetWellKnownGeogCS( 'WGS84' )
dst_ds.SetProjection( srs.ExportToWkt() )
raster = numpy.zeros( (512, 512) )
dst_ds.GetRasterBand(1).WriteArray(raster)
```

```python
from osgeo import gdal
import numpy
dataset = gdal.Open("geotiff_file.tif")
width = dataset.RasterXSize
height = dataset.RasterYSize
datas = dataset.ReadAsArray(0,0,width,height)
driver = gdal.GetDriverByName("GTiff")
tods = driver.Create("x_geotiff_file_3.tif",width,height,3,options=["INTERLEAVE=PIXEL"]) 
tods.WriteRaster(0,0,width,height,datas.tobytes(),width,height, band_list=[1,2,3])
```
