```python
import copy
a = [1, 2, 3, 4, ['a', 'b']]
b = a
c = copy.copy(a)
d = copy.deepcopy(a)
a.append(5)
a[4].append('c')
print(a, b, c, d, sep = "\n")
# 以下输出结果错误的是：

# A. a == [1,2, 3, 4, ['a', 'b', 'c'], 5] 
# B. b == [1,2, 3, 4, ['a', 'b', 'c'], 5]
# C. c == [1,2, 3, 4, ['a', 'b', 'c']] 
# D. d == [1,2, 3, 4, ['a', 'b', 'c']]
```

```python
# 月球上物体的体重是在地球上的16.5%，假如你在地球上每年增长0.5kg，编写程序输出未来十年你在地球和月球上的体重。
```

```python
#请补全程序  weight是地球上的体重
weight = float(input("请输入你在地球上的体重(kg): "))
for i in range(10):
    weight += 0.5

    print("On earth: {:.2f}kg".format(weight))
    print("On moon: {:.2f}kg".format(0.165*weight))
```
