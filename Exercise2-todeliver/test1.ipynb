```python
from  osgeo import ogr
import os
os.chdir("D:/learnpy/learnOpengis-1/Exercise3-todeliver")
inshp = "./data-Exercise3/sites.shp"
datasource = ogr.Open(inshp)
driver = datasource.GetDriver()
layer = datasource.GetLayer(0)
```

```python
layerdef = layer.GetLayerDefn()
for i in range(layerdef.GetFieldCount()):
    fielddef = layerdef.GetFieldDefn(i)
    print(fielddef.GetName())
    print(fielddef.GetTypeName())
    print(fielddef.GetWidth())
    print(fielddef.GetPrecision())
    print(fielddef.GetComment())
```

```python
# 当前要素的几何信息和几何要素坐标
for i in range(layer.GetFeatureCount()):
    feature = layer.GetFeature(i)
    print(feature)
```

```python
count = layer.GetFeatureCount()
print(count)
```

```python
layer.ResetReading()
for feature in layer:
    geom = feature.GetGeometryRef()
    # print(geom.GetGeometryName())
    # print(geom.GetPointCount())
    print(geom.GetX(), geom.GetY())
```

```python
with open("saves.txt", "w", encoding="utf-8") as file:
    # 写入表头
    file.write("id,x,y,cover\n")

    # 遍历所有要素
    for feature in layer:
        feature_id = feature.GetField("ID")
        cover = feature.GetField("COVER")

        geometry = feature.GetGeometryRef()
        x = geometry.GetX()
        y = geometry.GetY()

        file.write(f"{feature_id},{x},{y},{cover}\n")

print("写入完成：saves.txt")
```

```python
layer.ResetReading()
feature = layer.GetNextFeature()
file = open("../Exercise3-todeliver/data-Exercise3.txt", "w")
while feature:
    geom = feature.GetGeometryRef()
    f_id = feature.GetField("ID")
    f_cover = feature.GetField("COVER")
    geom = feature.GetGeometryRef()
    f_x = geom.GetX()
    f_y = geom.GetY()

    file.write(f"{f_id},{f_cover},{f_x},{f_y}\n")

    feature = layer.GetNextFeature()
file.close()

```
