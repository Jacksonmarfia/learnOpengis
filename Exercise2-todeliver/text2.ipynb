```python
import os
from osgeo import ogr
driver=ogr.GetDriverByName('ESRI Shapefile')
os.chdir("D:/learnpy/learnOpengis-1/Exercise3-todeliver")
inshp = "./data-Exercise3/sites.shp"
datasource = ogr.Open(inshp)
driver = datasource.GetDriver()
layer_in = datasource.GetLayer(0)
```

```python
driver=ogr.GetDriverByName('ESRI Shapefile')
outshp = "./data-Exercise3/trees.shp"
ds = driver.CreateDataSource(outshp)
layer_out=ds.CreateLayer('trees_poi',geom_type=ogr.wkbPoint)
```

```python
layer_defn = layer_in.GetLayerDefn()
for i in range(layer_defn.GetFieldCount()):
    field_defn = layer_defn.GetFieldDefn(i)
    layer_out.CreateField(field_defn)
    print(field_defn.GetName())
```

```python
featureDefn = layer_out.GetLayerDefn()
```

```python
print(featureDefn.GetName())
```

```python
layer_in.ResetReading()
feature_in = layer_in.GetNextFeature()
while feature_in:
    if feature_in.GetField('COVER') == 'trees':
        feature_out = ogr.Feature(featureDefn)

        feature_out.SetField('ID', feature_in.GetField('id'))
        feature_out.SetField('COVER', feature_in.GetField('COVER'))

        geometry = feature_in.GetGeometryRef()

        if geometry is not None:
                feature_out.SetGeometry(geometry.Clone())

        layer_out.CreateFeature(feature_out)
        
    feature_in = layer_in.GetNextFeature()
```

```python
layer_in.ResetReading()
feature_in = layer_in.GetNextFeature()
count = 0
while feature_in:
    if feature_in.GetField('COVER') == 'trees':
        print(feature_in)
        count += 1
    feature_in = layer_in.GetNextFeature()
print(f"Number of tree features: {count}")
```

```python
# 当前要素的几何信息和几何要素坐标
for i in range(layer_out.GetFeatureCount()):
    feature = layer_out.GetFeature(i)
    print(feature)
```
