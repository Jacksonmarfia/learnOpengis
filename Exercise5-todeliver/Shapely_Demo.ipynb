```python
from pyproj import Proj
albers=Proj('+proj=aea +lon_0=105 +lat_1=25 +lat_2=47 +ellps=krass')
crs=CRS.from_proj4('+proj=aea +lon_0=105 +lat_1=25 +lat_2=47 +ellps=krass')
albers_x,albers_y=albers(105,36)
albers_x,albers_y
```

```python
utm=Proj(proj='utm',zone=48, ellps='krass')
crs_new=CRS.from_proj4('+proj=utm +zone=48 +ellps=krass')
utm_x,utm_y=utm(105,36)
print(utm_x,utm_y)
```

```python
from pyproj import transform   #老版本，已经不推荐使用
to_utm_x,to_utm_y=transform(albers,utm,albers_x,albers_y)
print(to_utm_x,to_utm_y)
```

```python
from pyproj import Transformer
transformer=Transformer.from_crs(crs,crs_new)    #从 albers转换为 utm
transformer.transform(albers_x,albers_y)
```
